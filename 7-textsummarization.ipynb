```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
api_key=os.environ.get("GROQ_API_KEY")
llm=ChatGroq(groq_api_key=api_key,model="openai/gpt-oss-120b")
llm
```

```python
from langchain_core.messages import(
    AIMessage,
    HumanMessage,
    SystemMessage
)
```

```python
speech="""
India is moving toward a future where development is driven by innovation, self-reliance, technology, 
and the participation of its young people. The country must continue strengthening its manufacturing 
capabilities, supporting startups, improving infrastructure, and creating opportunities for every 
section of society. New technologies such as artificial intelligence can play an important role in 
improving public services, education, healthcare, and employment. At the same time, development should 
remain inclusive and sustainable, with greater attention to environmental protection and the empowerment 
of women and youth. The vision of a developed India requires citizens, businesses, institutions, and 
governments to work together with a shared sense of responsibility and purpose.
"""
```

```python
speech
```

```python
chat_messages = [
    SystemMessage(content="You are an expert with expertise in summarizing speeches."),
    HumanMessage(content=f"Please provide a short and concise summary of the following speech:\n Text: {speech}")]
```

```python
llm.get_num_tokens(speech)
```

```python
## getting the summary of the speech
llm.invoke(chat_messages).content
```

## Prompt Template for summarization

```python
from langchain_core.prompts import PromptTemplate

generic_template = """
Write a concise summary of the following speech:

Speech: {speech}

Translate the precise summary to {language} language.
"""

prompt = PromptTemplate(
    input_variables=["speech", "language"],
    template=generic_template
)

prompt
```

```python
complete_prompt=prompt.format(speech=speech, language="Hindi")
complete_prompt
```

```python
llm.get_num_tokens(complete_prompt)
```

```python
chain=prompt|llm
summary=chain.invoke({"speech":speech, "language":"Hindi"})
summary
```

```python
summary.content
```
