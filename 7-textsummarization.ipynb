## StuffDocumentChain Text Summarization

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("apjspeech.pdf")
docs = loader.load_and_split()
docs
```

```python
template="""
Write a concise and short summary of the following speech:
Speech :{text}
"""

prompt=PromptTemplate(input_variables=["text"], template=template)
prompt
```

```python
# Chain
chain = prompt | llm

# Combine PDF pages
speech = "\n".join(doc.page_content for doc in docs)

# Generate summary
summary = chain.invoke({
    "text": speech
})

print(summary.content)
```

## Map Reduce to Summarize Large Documents

```python
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

final_documents= RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=50).split_documents(docs)
final_documents
```

```python
len(final_documents)
```

```python
chunks_prompt="""
Please summarise the below speech:
Speech:`{text}`
Summary:
"""

map_prompt_template=PromptTemplate(input_variables=["text"], template=chunks_prompt)
map_prompt_template
```

```python
final_prompt="""
Provide the final summary of the entire speech with these important points.
Add a motivation title, start the precise summary with an introduction and provide the summary 
in number points for the speech.
Speech:{text}

"""
final_prompt_template=PromptTemplate(input_variables=["text"], template=final_prompt)
```

```python
map_chain = map_prompt_template | llm
```

```python
chunk_summaries = []

for doc in final_documents:
    response = map_chain.invoke({
        "text": doc.page_content
    })
    chunk_summaries.append(response.content)
```

```python
print(chunk_summaries)
```

```python
# Combine summaries
combined_summaries = "\n\n".join(chunk_summaries)
```

```python
# REDUCE CHAIN
final_chain = final_prompt_template | llm
```

```python
# Final summary
final_summary = final_chain.invoke({
    "text": combined_summaries
})
```

```python
final_summary
```

```python
final_summary.content
```

### Refine Chain for Summarization

```python
summary = map_chain.invoke({
    "text": final_documents[0].page_content
}).content

for doc in final_documents[1:]:
    summary = final_chain.invoke({
        "existing_summary": summary,
        "text": doc.page_content
    }).content

print(summary)
```
